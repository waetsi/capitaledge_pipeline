### Save Data

```python
import os
from datetime import datetime
import json

# Validate data first
if "Time Series (Daily)" in data:
    print("Valid data received ✅")

    # Create folder if it doesn't exist
    os.makedirs("data/raw", exist_ok=True)

    # Timestamp
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    # File path
    file_path = f"data/raw/{symbol}_{timestamp}.json"

    # Save file
    with open(file_path, "w") as f:
        json.dump(data, f, indent=4)

    print(f"Data saved to {file_path}")

else:
    print("Warning: API response may be incomplete ⚠️")
```

### Data Storage Summary

The extracted data has been successfully stored in the raw data layer (`data/raw/`) using a timestamped filename.

This approach ensures:
- Data traceability
- Historical record keeping
- Readiness for incremental data loading in subsequent stages
